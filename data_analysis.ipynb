```python
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import squarify

mydata = pd.read_csv("age_salary_hours.csv", index_col=0)

palette = sns.color_palette('colorblind', n_colors=mydata['Education'].nunique())

fig = plt.figure(figsize=(12, 14))
gs = fig.add_gridspec(3, 1, height_ratios=[1, 2, 2])

ax1 = fig.add_subplot(gs[0])
edu_counts = mydata['Education'].value_counts()
squarify.plot(sizes=edu_counts.values, label=edu_counts.index, ax=ax1, color=palette, alpha=0.8)
ax1.set_title('Degrees', fontsize=16)
ax1.axis('off')

ax2 = fig.add_subplot(gs[1])
sns.violinplot(x='Education', y='Annual Salary', data=mydata, ax=ax2, 
               palette=palette, hue='Education', legend=False)
ax2.set_xlabel('Education')
ax2.set_ylabel('Annual Salary')
plt.setp(ax2.get_xticklabels(), rotation=15, ha='right')

ax3 = fig.add_subplot(gs[2])
sns.violinplot(x='Education', y='Weekly hours', data=mydata, ax=ax3, 
               palette=palette, hue='Education', legend=False)
ax3.set_xlabel('Education')
ax3.set_ylabel('Weekly hours')
plt.setp(ax3.get_xticklabels(), rotation=15, ha='right')

plt.tight_layout()
plt.show()
```
